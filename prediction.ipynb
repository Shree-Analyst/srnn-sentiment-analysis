### Prediction with Simple RNN

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import sequence
```

```python
## Load word indices & dataset
word_index = imdb.get_word_index()
reverse_word_index = {value: key for key, value in word_index.items()}
```

```python
## Load model
model = load_model('simple_rnn_imdb.keras')
model.summary()
```

```python
model.get_weights()
```

```python
## Helper Functions

def decode_review(review): # Decoding reviews
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in review])

def preprocess_text(text): # Preprocess text
    words = text.lower().split()
    encoded_review = [word_index.get(word, 2) + 3 for word in words]
    padded_review = sequence.pad_sequences([encoded_review], maxlen = 500)
    return padded_review
```

```python
## Prediction Function

def predict_sentiment(review):
    padded_input = preprocess_text(review)
    prediction = model.predict(padded_input)

    sentiment = 'positive' if prediction[0][0] > 0.5 else 'negative'

    return sentiment, prediction[0][0]
```

```python
## Test Prediction

sample_review = 'This movie was fantastic! The acting was great and the plot was thrilling.'
sentiment, score = predict_sentiment(sample_review)
print(f'Sentiment: {sentiment} \nScore: {score}')
```
