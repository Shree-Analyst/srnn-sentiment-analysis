```python
## Sentences

sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
     'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good']

sent
```

```python
## One Hot Representation

from tensorflow.keras.preprocessing.text import one_hot

vocab_size = 10000 # Define vocabulary size
ohr = [one_hot(docs, vocab_size) for docs in sent]
ohr
```

## Word Embedding Representation

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
## Feature Engineering

sent_length = 8
embedded_docs = pad_sequences(ohr, padding = 'pre', maxlen = sent_length)
print(embedded_docs)
```

```python
## Feature Representation

dim = 10

model = Sequential()
model.add(Embedding(vocab_size, dim, input_length = sent_length))
model.compile('adam', 'mse')
model.build(input_shape = (None, sent_length))
model.summary()
```

```python
print(sent[0:2])
print(embedded_docs[0:2])
model.predict(embedded_docs)[0][0:2]
```
